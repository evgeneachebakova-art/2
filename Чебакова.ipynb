3) Составить программу, осуществляющую перевод величин из радианной меры в градусную или наоборот. Программа должна запрашивать, какой перевод нужно осуществить, и выполнять указанное действие.

```python
import math
a = input("Выберите действие 1(перевести радианы в градусы) или 2(перевести градусы в радианы) ")
if a == "1":
    b = float(input("Ведите значение в радианах: "))
    c = math.degrees(b)
    print(f'{b} рад = {c} гр')
elif a == "2":
    b = float(input("Ведите значение в градусах: "))
    c = math.radians(b)
    print(f'{b} гр = {c} рад')
else :
    print("Ошибка: нужно было ввести 1 или 2")
```

4)	 Дано целое число N и набор из N целых чисел. Если в наборе имеются положительные числа, то вывести True; в противном случае вывести False.

```python
n = int(input("N: "))
numbers = list(map(int, input("Ведите числа в одну строку через пробел ").split()))
if len(numbers) != n :
    print("Ошибка: чесел не N")
else:
    for x in numbers:
        if x > 0:
            print(True)
            break
    else:
            print(False)
```

1.	Найти все простые натуральные числа, не превосходящие n, двоичная запись которых представляет собой палиндром, т.е. читается одинаково слева направо и справа налево.

```python
n = int(input("n: "))
for x in range(2,n+1):
    a = bin(x)[2:]
    if a != a[::-1]:
        continue
    b = True
    i = 2
    while i*i <= x :
        if x % i == 0 :
            b = False
            break
        i += 1
    if b == True :
        print(x, end=" ")
```

2.	Задан массив D из 6 элементов. Определить следующие суммы: D[1] + D[2] + D[3]; D[3] + D[4] + D[5]; D[4] + D[5] + D[6]. Пояснение. Составить функцию вычисления суммы трех последовательно расположенных элементов массива с номерами от k до m .

```python
def sum_free(arr, k) :
    return arr[k-1]+arr[k]+arr[k+1]
D = list(map(float, input("Введите 6 элементов массива через пробел: ").split()))
s1 = sum_free(D, 1)
s2 = sum_free(D, 3)
s3 = sum_free(D, 4)
print(f'D[1]+D[2]+D[3] = {s1} \nD[3]+D[4]+D[5] = {s2} \nD[4]+D[5]+D[6] = {s3}')
```

```python
def factorial(k):
    p = 1
    for i in range(2,k+1):
        p *= i
    return p
n = int(input("n(четное, <= 20): "))
if n % 2 != 0 or n > 20 or n < 2:
    print("Ошибка: n должно быть четным, от 2 до 20")
else:
    s = 0
    for k in range(2,n+1,2):
        s += factorial(k)
    print(s)
```
