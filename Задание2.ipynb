Задача 13: Определите класс Contact с полями имя, номер телефона и электронная почта. Реализуйте методы update_phone(new_phone), который обновляет номер телефона, и display_info(), который выводит информацию о контакте.

```python
class Contact :

    def __init__(self, name, phone, email):
        self.name = name
        self.phone = phone
        self.email = email

    def update_phone(self, new_phone):
        self.phone = new_phone
        print("Успешно обновлено")

    def display_info(self):
        print(f'Имя: {self.name}, номер телефона: {self.phone}, электронная почта: {self.email}')

# Пример использования класса
if __name__ == "__main__":
    my_contact = Contact("Jenny", "+9625624052", "evgeneachebakova@mail.ru")
    my_contact.display_info()
    my_contact.update_phone("+9625624072")
    my_contact.display_info()
```

Задача 13. Определите класс Contact, который наследуется от класса Profile. Класс Contact должен добавлять поля номер телефона и электронная почта, а также методы для обновления номера телефона.

```python
class Profile :
    def __init__(self, name):
        self.name = name

class Contact(Profile):
    def __init__(self, name, phone, email):
        super().__init__(name)
        self.phone = phone
        self.email = email

    def update_phone(self, new_phone):
        self.phone = new_phone
        print("успешно обновлено")

    def display_info(self):
        print(f'Имя: {self.name}, номер телефона: {self.phone}, электронная почта: {self.email}')

if __name__ == "__main__":
    my_contact = Contact("Jenny", "+9625624052", "evgeneachebakova@mail.ru")
    my_contact.display_info()
    my_contact.update_phone("+9625624072")
    my_contact.display_info()
```
