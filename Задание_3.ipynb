```python
class Creature:
    def get_class(self):
        return 'Creature'

class Mammal(Creature):
    def get_class(self):
        return super().get_class() + " , Mammal"

class Bird(Creature):
    def get_class(self):
        return 'Bird'

creature = Creature()
mammal = Mammal()
bird = Bird()

print(creature.get_class())
print(mammal.get_class())
print(bird.get_class())

"""
Что будет, если в наследнике не переопределять метод?
Наследник просто унаследует метод родителя и будет использовать его.
Например, если в Mammal не писать свой get_class(), то вызов
mammal.get_class() вернёт "Creature" — как у базового класса.
Как вызвать родительскую версию через super()?
Внутри метода наследника можно написать super().get_class().
Это вызовет метод из базового класса.
"""
```

Задача 6 (статические методы). Зоопарку нужна утилита для проверки номеров вольеров, которая не зависит от конкретного объекта и может вызываться сама по себе.
```python
class ZooUtils:
    @staticmethod
    def is_valid_enclosure(number):
        if 1 <= number <= 201:
            return True
        else:
            return False

print(ZooUtils.is_valid_enclosure(42))
print(ZooUtils.is_valid_enclosure(242))

"""
Чем статический метод отличается от обычного?
Обычный метод (с self) работает с конкретным объектом и имеет доступ
к его полям. Статический метод не принимает self, он работает как
обычная функция, просто лежит внутри класса для организации кода.
Когда использовать статический метод?
Когда функция логически связана с классом, но не нуждается в доступе
к данным объекта или класса.
"""
```

Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и *args. Создайте класс FeedDesk с методом feed(animal, food=None, time=None), который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Зарегистрировано кормление животного: », «Животное кормят: » или «Животное кормят в ».
```python
class FeedDesk:
    def feed(self, animal, food=None, time=None):
        if food is None and time is None:
            print(f"Зарегистрировано кормление животного: {animal}")
        elif food is not None and time is None:
           print(f"Животное кормят: {animal} — {food}")
        else:
            print(f"Животное кормят в {time}: {animal} — {food}")

feeddesk = FeedDesk()
feeddesk.feed("Симба")
feeddesk.feed("Симба", "Мясо")
feeddesk.feed("Симба", "Мясо", "15:20")

"""
Почему в Python имитируют перегрузку через значения по умолчанию?
Потому что в Python метод — это один объект-функция. Нельзя определить
несколько методов с одинаковым именем, так как второй перезапишет первый.
Значения по умолчанию позволяют одному методу работать в разных режимах:
если аргумент не передан, он равен None, и мы проверяем это внутри метода.
"""
```

Задача 11 (магические методы __str__ и __repr__). Животных часто выводят на экран — в каталогах зоопарка, отчётах, табличках у вольеров. По умолчанию print(animal) выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс Animal магический метод __str__, который возвращает строку вида «Животное "Симба" (лев, 7 лет)». Убедитесь, что при вызове print(animal) выводится именно эта строка. Дополнительно реализуйте метод __repr__, который возвращает техническое представление объекта, например Animal('Симба', 'лев', 7) — оно используется при выводе списка животных и в отладчике. Поясните разницу между __str__ и __repr__ и когда какой используется: __str__ — для человека, __repr__ — для разработчика.

```python
class Animal:
    def __init__(self, name, species, age):
        self.name = name
        self.species = species
        self.age = age
    def __str__(self):
        return f'Животное "{self.name}" ({self.species}, {self.age} лет)'
    def __repr__(self):
        return f"Animal('{self.name}', '{self.species}', {self.age})"

animal = Animal("Симба", "лев", 8)
print(animal)
print(str(animal))
print(repr(animal))

animals = [Animal("Симба", "лев", 7), Animal("Дамбо", "слон", 10)]
print(animals)

"""
Разница между __str__ и __repr__:
__str__ — для человека, вызывается через print() и str(). Возвращает
красивое, понятное представление объекта.

__repr__ — для разработчика, вызывается в списках, в отладчике, через repr().
Возвращает техническое представление, по которому можно понять, как создать объект.

Если __str__ не определён, Python использует __repr__ вместо него.
""
```

Объясните в комментариях, почему при переопределении __eq__ нужно также переопределять __hash__ — что произойдёт, если этого не сделать.

```python
class Animal:
    def __init__(self, name, species, age):
        self.name = name
        self.species = species
        self.age = age

    def __eq__(self, other):
        if not isinstance(other, Animal):
            return False
        elif self.name == other.name and self.species == other.species:
            return True

    def __hash__(self):
        return hash((self.name, self.species))

animal1 = Animal("Симба", "лев", "8")
animal2 = Animal("Симба", "лев", "7")
animal3 = Animal("Дамбо", "слон", "10")

print(animal1 == animal2)
print(animal1 == animal3)

zoo = {animal1, animal2, animal3}
print("Размер множества:", len(zoo))

ages = {animal1: "малыш", animal3: "взрослый"}
print(ages[animal2])

"""
Почему нужно переопределять __hash__ вместе с __eq__?
Если два объекта равны (==), у них должен быть одинаковый хеш.
Если переопределить только __eq__, Python обнулит __hash__ автоматически и объект нельзя будет положить в set или dict.
"""
```

Задача 13 (магические методы __add__ и __len__). В зоопарке есть секции с животными, и иногда нужно объединять их (например, при объединении секторов) или узнавать, сколько животных в секции. Создайте класс Section (секция), в котором хранится список животных. Реализуйте магический метод __add__ для объединения двух секций в одну (возвращается новая секция со всеми животными) и метод __len__, возвращающий количество животных. Продемонстрируйте работу: создайте две секции, объедините их через s3 = s1 + s2 и выведите len(s3). Объясните, почему __add__ возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

```python
class Animal:
    def __init__(self, name, species, age):
        self.name = name
        self.species = species
        self.age = age

class Section:
    def __init__(self, animals = None):
        if animals is None:
            self.animals = []
        else:
            self.animals = animals

    def __add__(self, other):
        animals3 = self.animals + other.animals
        return Section(animals3)

    def __len__(self):
        return len(self.animals)

animal1 = Animal("Симба", "лев", 7)
animal2 = Animal("Дамбо", "слон", 10)
animal3 = Animal("Жираф", "жираф", 5)

s1 = Section([animal1, animal2])
s2 = Section([animal3])

s3 = s1 + s2

print(f"В s1: {len(s1)} животных")
print(f"В s2:  {len(s2)} животных")
print(f"В s3: {len(s3)} животных")

"""
Почему __add__ возвращает новый объект?
Если бы __add__ менял существующий объект, код стал бы непредсказуемым:
s1 + s2 неожиданно изменил бы s1, что ломает логику.
Как магические методы делают объекты похожими на встроенные типы?
Они позволяют использовать стандартные операторы (+, len, ==) и функции
для объектов. Section ведёт себя как список: её можно сложить
через + и измерить через len().
"""
```

Задача 14 (свойство @property). В зоопарке важно, чтобы возраст животного был корректным — нельзя допустить отрицательное значение или значение больше 100. Перепишите класс Animal так, чтобы поле age было реализовано через свойство @property. Проверки: возраст должен быть целым числом и лежать в диапазоне от 0 до 100 включительно. При нарушении должно выбрасываться исключение ValueError с понятным сообщением. Покажите, что обращение к animal.age работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: animal.age = -1 → ошибка, animal.age = 200 → ошибка, animal.age = "abc" → ошибка, animal.age = 7 → успех. Объясните, чем @property удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

```python
class Animal:
    def __init__(self, name, species, age):
        self.name = name
        self.species = species
        self.age = age

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, value):
        if not isinstance(value, int):
            raise ValueError("Возраст должен быть целым числом")
        if 0 <= value <= 100:
            self._age = value
        else:
            raise ValueError("Возраст должен быть от 0 до 100")

animal = Animal("Симба", "Лев", 8)
print(f'Возраст: {animal.age}')
animal.age = 9
print(f'Возраст: {animal.age}')

try:
    animal.age = -1
except ValueError as e:
    print("Ошибка:", e)

try:
    animal.age = 200
except ValueError as e:
    print("Ошибка:", e)

try:
    animal.age = "abc"
except ValueError as e:
    print("Ошибка:", e)

"""
Чем @property удобнее геттеров/сеттеров?
С @property обращение выглядит как работа с обычным полем: animal.age = 7.
Не нужно писать animal.set_age(7) и animal.get_age().
Что такое «вычисляемый атрибут»?
Это свойство, значение которого не хранится, а вычисляется каждый раз
при обращении.
"""
```

Задача 15 (классовый метод @classmethod). В зоопарк часто поступают данные о животных в виде строк — например, из файла или базы данных: "Симба;лев;7". Добавьте в класс Animal классовый метод from_string(cls, data), который создаёт объект животного из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание животного через этот метод и вывод результата. Объясните разницу между @classmethod, @staticmethod и обычным методом: обычный метод получает первым аргументом экземпляр (self) и работает с конкретным объектом, @classmethod получает сам класс (cls) и часто используется как альтернативный конструктор, @staticmethod не получает ни self, ни cls и представляет собой просто утилиту. Расскажите, когда какой использовать.

```python
class Animal:
    def __init__(self, name, species, age):
        self.name = name
        self.species = species
        self.age = age

    @classmethod
    def from_string(cls, data):
        parts = data.split(";")
        name = parts[0]
        species = parts[1]
        age = int(parts[2])
        return cls(name, species, age)

animal = Animal.from_string("Симба;лев;7")
print(f"Имя: {animal.name}, \nВид: {animal.species}, \nВозраст: {animal.age} лет")

"""
Разница между обычным методом, @classmethod и @staticmethod.
Обычный метод работает с конкретным объектом. Ему важно знать,
с кем он имеет дело, чтобы прочитать или изменить его личные данные.
Классовый метод (@classmethod) работает с классом в целом, а не с конкретным объектом. Он часто
используется как альтернативный конструктор, когда нужно создать объект
нестандартным способом, например, из строки.
Статический метод (@staticmethod) это функция, которая логически связана с классом, но не нуждается ни в данных
объекта, ни в атрибутах класса.

Выбор метода зависит от задачи. Если необходимо работать с данными конкретного объекта,
используется обычный метод. Если нужно создать объект нестандартным
способом или обратиться к атрибутам класса, используется @classmethod.
Если функция не нуждается ни в объекте, ни в классе используется @staticmethod.
"""
```

Задача 16 (классовые vs экземплярные атрибуты). Зоопарку нужно вести общий счётчик всех животных — сколько всего животных зарегистрировано в системе.
```python
class Animal:
    count = 0
    def __init__(self, name, species, age):
        self.name = name
        self.species = species
        self.age = age
        Animal.count += 1

animal1 = Animal("Симба", "лев", 7)
animal2 = Animal("Дамбо", "слон", 10)
animal3 = Animal("Жираф", "жираф", 5)

print("Всего животных:", Animal.count)

animal1.count = 100
print("Всего животных:", Animal.count)
print("animal1.count", animal1.count)

"""
Разница между классовыми и экземплярными атрибутами:
Классовый атрибут (например, count = 0) — общий для всех объектов класса.
Он хранится в самом классе, и для всех объектов это одно и то же значение.
Если изменить классовый атрибут через класс, это
изменение отразится на всех объектах.
Экземплярный атрибут (например, self.name) — свой у каждого объекта.
Он хранится в конкретном объекте и не влияет на другие объекты.

В каком порядке Python ищет атрибут при obj.attr:
1. Сначала ищет в самом объекте (экземплярные атрибуты)
2. Если не нашёл — ищет в классе (классовые атрибуты)
3. Если и там нет — ищет в родительских классах (наследование)
Поэтому если написать a1.count = 100, Python создаст экземплярный атрибут
count у объекта a1, который "перекроет" классовый count. Но Animal.count
останется прежним — мы просто создали новый атрибут у конкретного объекта.
"""
```

```python
class Zoo():
    def __init__(self, name):
        self.name = name
        self.animals = []

    def add_animal(self, animal):
        self.animals.append(animal)
        animal.zoo = self

class Animal:
    def __init__(self, name, species):
        self.name = name
        self.species = species
        self.zoo = None

    def info(self):
        if self.zoo is not None:
            return f"{self.name} ({self.species}) живёт в зоопарке '{self.zoo.name}'"
        else:
            return f"{self.name} ({self.species}) пока без зоопарка"

class Keeper:
    def __init__(self, name):
        self.name = name
        self.animals = []

    def add_animal(self, animal):
        self.animals.append(animal)

    def feed_all(self):
        for animal in self.animals:
            print(f"{self.name} кормит {animal.name}")

zoo = Zoo("Zoo1")

animal1 = Animal("Симба", "лев")
animal2 = Animal("Дамбо", "слон")

zoo.add_animal(animal1)
zoo.add_animal(animal2)

keeper = Keeper("Keeper1")
keeper.add_animal(animal1)
keeper.add_animal(animal2)

print(animal1.info())
keeper.feed_all()

keeper = None
print("Животных в зоопарке:", len(zoo.animals))
print(animal1.info())

"""
Разница между композицией, агрегацией и ассоциацией заключается в силе
связи между объектами. Ассоциация — это самая слабая связь, при которой
объекты просто знают о существовании друг друга, но остаются полностью
независимыми (например, животное имеет ссылку на зоопарк, но может
существовать и без него). Агрегация представляет собой более тесную,
но всё ещё мягкую связь типа «содержит»: один объект включает в себя
другие, однако эти части могут жить отдельно. Композиция же создаёт
самую жёсткую связь типа «является частью»: часть не может существовать
без целого. Как в случае с вольером, который создаётся внутри животного:
если удалить животное, его вольер тоже перестанет существовать, поскольку
у него нет независимой жизни.
"""
```

```python
class Keeper:
    def get_role(self):
        return f"смотритель"

class Vet:
    def get_role(self):
        return f"ветеринар"


class  KeeperVet(Keeper, Vet):
    pass

keepervet = KeeperVet()
print("Роль сотрудника:", keepervet.get_role())

print("\nПорядок разрешения методов (MRO):")
for cls in KeeperVet.__mro__:
    print("-", cls.__name__)


"""
MRO (Method Resolution Order) — это порядок, в котором Python ищет
методы при наследовании. При множественном наследовании Python использует
алгоритм C3-линеаризации, который выстраивает строгую линейную очередь
поиска: сначала сам класс, затем родители слева направо, и в конце базовый
класс object.

В классе KeeperVet(Keeper, Vet) при вызове get_role() будет использован
метод из класса Keeper. Это происходит потому, что Keeper указан первым
в списке наследования, и Python находит метод там, прекращая дальнейший
поиск.

«Проблема ромба» возникает, когда два родительских класса наследуются от
одного общего предка и переопределяют его метод, а дочерний класс наследуется
от обоих родителей. Неясно, метод какого родителя использовать. Python
решает эту проблему с помощью алгоритма C3, который гарантирует, что каждый
класс в иерархии проверяется только один раз.
"""
```

Подключите миксин к классу Animal через множественное наследование, причём миксин должен идти первым в списке родителей: class Animal(LoggableMixin): .... Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

```python
from dataclasses import dataclass
from datetime import datetime

class LoggableMixin:
    def log(self, message):
        now = datetime.now().strftime("%H:%M:%S")
        print(f"[{self.__class__.__name__}] {now}: {message}")

@dataclass
class Animal(LoggableMixin):
    name: str
    species: str
    age: int

animal = Animal("Симба", "лев", 7)
print(animal)
animal.log("Покормили животное")

animal2 = Animal("Симба", "лев", 7)
print("Равны?", animal == animal2)

"""
Миксин — это класс, который добавляет одну узкую функциональность
к другому классу через множественное наследование. В отличие от обычного
родительского класса, миксин не предназначен для самостоятельного
использования. Он не создаёт полноценные объекты и обычно не имеет
собственного конструктора или состояния, а дает
другим классам дополнительную способность. Миксин подключается как первый родитель
в списке наследования, чтобы его методы имели приоритет при поиске (MRO).
"""
```

```python
import sys

# Класс БЕЗ __slots__ (для сравнения)
class AnimalNormal:
    def __init__(self, name, species, age):
        self.name = name
        self.species = species
        self.age = age

# Класс С __slots__
class AnimalSlotted:
    __slots__ = ("name", "species", "age")

    def __init__(self, name, species, age):
        self.name = name
        self.species = species
        self.age = age

normal = AnimalNormal("Симба", "лев", 7)
slotted = AnimalSlotted("Симба", "лев", 7)

try:
    slotted.weight = 200
    print("slotted.weight =", slotted.weight)
except AttributeError as e:
    print("Ошибка slotted:", e)

print("\nРазмер оболочки normal:", sys.getsizeof(normal), "байт")
print("Размер оболочки slotted:", sys.getsizeof(slotted), "байт")

from pympler import asizeof
print("Полный размер normal:", asizeof.asizeof(normal))
print("Полный размер slotted:", asizeof.asizeof(slotted))

"""
Как __slots__ экономит память?
По умолчанию Python хранит атрибуты объекта в словаре __dict__, который
занимает много места (около 100+ байт даже для пустого словаря). Когда мы
определяем __slots__, Python не создаёт __dict__, а выделяет фиксированное
место только под указанные атрибуты.
Что теряется при использовании __slots__?
Нельзя динамически добавлять новые атрибуты объекту — только те, что
указаны в __slots__. Отсутствует __dict__, поэтому не работают функции
vars() и obj.__dict__. Наследование усложняется: дочерний класс должен
тоже определять __slots__, иначе у него появится __dict__.
В каких случаях стоит применять __slots__?
Когда создаются юольшое колличество объектов с фиксированным набором
атрибутов.
"""
```

Задача 22 (дескрипторы). Свойство @property — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор RangeDescriptor(min_value, max_value), который хранит значение в instance.__dict__ (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая ValueError при нарушении. Используйте этот дескриптор в классе Animal для полей age (например, диапазон 0–100) и weight (диапазон 1–10000), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» @property, зачем нужен метод __set_name__ и в чём преимущество дескриптора перед @property, если одинаковая проверка повторяется в десяти полях.

```python
class RangeDescriptor:
    def __init__(self, min_value, max_value):
        self.min_value = min_value
        self.max_value = max_value

    def __set_name__(self, owner, name):
        self.name = name

    def __get__(self, instance, owner):
        if instance is None:
            return self
        return instance.__dict__.get(self.name)

    def __set__(self, instance, value):
        if not isinstance(value, (int, float)):
             raise ValueError(f"{self.name} должно быть числом")
        if not (self.min_value <= value <= self.max_value):
             raise ValueError(
                 f"{self.name} должно быть от {self.min_value} до {self.max_value}")
        instance.__dict__[self.name] = value


class Animal:
    age = RangeDescriptor(0, 100)
    weight = RangeDescriptor(1, 10000)

    def __init__(self, name, species, age, weight):
        self.name = name
        self.species = species
        self.age = age
        self.weight = weight

animal = Animal("Симба", "лев", 7, 200)
print(f"Возраст: {animal.age}")
print(f"Вес: {animal.weight}")

try:
    animal.age = -5
except ValueError as e:
    print("Ошибка:", e)

try:
    animal.weight = 50000
except ValueError as e:
    print("Ошибка:", e)

animal2 = Animal("Дамбо", "слон", 10, 5000)
print(f"\nСимба: возраст={animal.age}, вес={animal.weight}")
print(f"Дамбо: возраст={animal2.age}, вес={animal2.weight}")

"""
Как дескрипторы работают «под капотом» @property?
Сам декоратор @property реализован через механизм дескрипторов. Когда используется @property, Python создаёт специальный объект-дескриптор, у которого
есть методы __get__, __set__ и __delete__.
Зачем нужен метод __set_name__?
Когда Python создаёт класс, он автоматически вызывает __set_name__ для
каждого дескриптора, передавая ему имя, под которым дескриптор
зарегистрирован в классе.Без __set_name__ пришлось бы
передавать имя вручную при создании дескриптора.
В чём преимущество дескриптора перед @property при повторении логики?
С @property для десяти полей с одинаковой проверкой пришлось бы написать
двадцать методов (геттер и сеттер для каждого). С дескриптором логика
пишется один раз в классе RangeDescriptor и переиспользуется для всех
полей: age = RangeDescriptor(0, 100), weight = RangeDescriptor(1, 10000).
"""
```

Задача 23 (метаклассы). В зоопарке должен быть только один реестр животных — если создать второй, данные могут разойтись. Создайте метакласс SingletonMeta, который гарантирует, что у класса Zoo будет только один экземпляр. Проверьте, что Zoo() дважды возвращает один и тот же объект (сравните через is). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

```python
class SingletonMeta(type):
    _instances = {}

    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]

class Zoo(metaclass=SingletonMeta):
    def __init__(self, name="Зоопарк"):
        self.name = name
        self.animals = []

    def add_animal(self, animal_name):
        self.animals.append(animal_name)

zoo1 = Zoo("ырл")
zoo2 = Zoo("рар")

print("zoo1 is zoo2:", zoo1 is zoo2)  # True!
print("zoo1.name:", zoo1.name)        # "Московский зоопарк"
print("zoo2.name:", zoo2.name)        # Тоже "Московский зоопарк"!

zoo2.add_animal("Симба")
print("zoo1.animals:", zoo1.animals)

"""
Метакласс — это «класс для классов»: он управляет созданием самих классов,
точно так же, как обычный класс управляет созданием объектов. По умолчанию
все классы в Python создаются метаклассом type. В повседневных проектах
метаклассы пишут редко, но они используются в крупных фреймворках (Django,
SQLAlchemy) для решения задач вроде автоматической регистрации классов,
валидации атрибутов или реализации паттерна Singleton, как в этой задаче.
"""
```

Задача 24 (паттерн Singleton). Реализуйте класс AnimalRegistry через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода __new__ (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; __new__ проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

```python
# Через метакласс (как в задаче 23)
class SingletonMeta(type):
    _instances = {}
    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]

class Animal1(metaclass=SingletonMeta):
    def __init__(self):
        self.animals = []

    def add(self, animal):
        self.animals.append(animal)

# Через __new__ (без метакласса)
class Animal2:
    _instance = None

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

    def __init__(self):
        if not hasattr(self, '_initialized'):
            self.animals = []
            self._initialized = True

    def add(self, animal):
        self.animals.append(animal)

# Способ через метакласс
r1 = Animal1()
r2 = Animal1()
print("r1 is r2:", r1 is r2)
r1.add("Симба")
print("r2.animals:", r2.animals)

# Способ через __new__
r3 = Animal2()
r4 = Animal2()
print("r3 is r4:", r3 is r4)
r3.add("Дамбо")
print("r4.animals:", r4.animals)

"""
Singleton полезен для кэша, конфигурации,
логгера или реестра — везде, где нужен единственный экземпляр с общим
состоянием. Однако он вреден при тестировании (состояние сохраняется
между тестами), скрывает зависимости и мешает подменять реализации —
в таких случаях лучше использовать внедрение зависимостей.
"""
```
